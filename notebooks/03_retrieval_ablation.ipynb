# 03: Retrieval Ablation
Compare Dense Search vs Sparse Search (BM25) vs Hybrid Search vs Hybrid + Reranker.

```python
import sys
sys.path.append('..')

from config.settings import get_settings
from src.pipeline.rag_pipeline import PinnacleRAGPipeline

settings = get_settings()
pipeline = PinnacleRAGPipeline(settings)
pipeline.ingest() # Make sure data is ingested
```

```python
# Define a test query
query = "What is the main architecture of this system?"

# 1. Base Retrieval (Dense + Sparse Hybrid without Reranker)
retriever = pipeline.get_retriever()
base_results = retriever.retrieve(query, top_k=5)

print("=== Base Hybrid Results ===")
for doc in base_results:
    print(f"- {doc.page_content[:100]}...")
```

```python
# 2. Reranked Results (Hybrid + Cross-Encoder)
reranker = pipeline._reranker
reranked_results = reranker.rerank(query, base_results, top_k=3)

print("\n=== Reranked Results ===")
for doc in reranked_results:
    print(f"- {doc.page_content[:100]}...")
```
