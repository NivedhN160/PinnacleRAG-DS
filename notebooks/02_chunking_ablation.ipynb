# 02: Chunking Ablation
In this notebook, we experiment with different chunk sizes and overlaps to understand their impact on the resulting chunk distributions.

```python
import sys
import copy
import pandas as pd
import matplotlib.pyplot as plt

sys.path.append('..')

from config.settings import get_settings
from src.ingestion.loader import DocumentLoader
from src.ingestion.cleaner import TextCleaner
from src.ingestion.chunker import DocumentChunker

settings = get_settings()
```

```python
# Load and Clean Docs
loader = DocumentLoader(settings)
raw_docs = loader.load_directory(settings.raw_data_path)
cleaner = TextCleaner(settings)
docs = cleaner.clean_batch(raw_docs)
print(f"Processing {len(docs)} documents.")
```

```python
# Test different chunk sizes
chunk_configs = [
    (500, 50),
    (1000, 100),
    (1500, 150)
]

results = []

for size, overlap in chunk_configs:
    # Override settings
    cfg = copy.deepcopy(settings)
    cfg.chunk_size = size
    cfg.chunk_overlap = overlap
    
    chunker = DocumentChunker(cfg)
    chunks = chunker.chunk_documents(docs)
    
    stats = chunker.get_chunk_stats(chunks)
    stats['config'] = f"{size}/{overlap}"
    results.append(stats)

df_results = pd.DataFrame(results)
display(df_results)
```
